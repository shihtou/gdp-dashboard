```python
# 套件載入 ------------------------------------------------------------

# 清除工作環境
rm(list = ls())

# 載入研究所需套件
library(readxl)
library(stringr)
library(zoo)
library(urca)
library(mgcv)


# 資料匯入與欄位設定 --------------------------------------------------

# 匯入 Excel 資料
data <- read_excel("論文數據.xlsx")

# 重新命名欄位
names(data) <- c(
  "ym",
  "price",
  "CPI",
  "M2",
  "BL",
  "RATE"
)

# 檢查資料結構
head(data)
str(data)
summary(data)


# 日期變數與應變數建立 ------------------------------------------------

# 擷取民國年份
data$year_m <- as.numeric(
  str_extract(data$ym, "^\\d+")
)

# 擷取月份
data$month <- as.numeric(
  str_extract(data$ym, "(?<=年)\\d+(?=月)")
)

# 民國年轉換為西元年
data$year <- data$year_m + 1911

# 建立年月日期變數
data$date <- as.yearmon(
  paste(data$year, data$month),
  format = "%Y %m"
)

# 依日期排序
data <- data[order(data$date), ]

# 重新設定列號
rownames(data) <- NULL

# 建立時間趨勢變數
data$time <- seq_len(nrow(data))

# 房價取自然對數
data$ln_price <- log(data$price)

# 檢查處理後資料
head(data)
summary(data)


# 遺漏值與描述性統計檢查 ----------------------------------------------

# 檢查各欄位遺漏值數量
colSums(is.na(data))

# 保留模型所需之完整觀測值
model_data <- na.omit(
  data[, c(
    "date",
    "time",
    "price",
    "ln_price",
    "RATE",
    "M2",
    "CPI",
    "BL"
  )]
)

# 描述性統計
summary(
  model_data[, c(
    "price",
    "ln_price",
    "RATE",
    "M2",
    "CPI",
    "BL"
  )]
)


# 時間序列趨勢圖 ------------------------------------------------------

# 設定圖形版面
par(mfrow = c(3, 2))

# 住宅成交單價
plot(
  model_data$date,
  model_data$price,
  type = "l",
  xlab = "年月",
  ylab = "住宅成交單價",
  main = "住宅成交單價時間序列圖"
)

# 重貼現率
plot(
  model_data$date,
  model_data$RATE,
  type = "l",
  xlab = "年月",
  ylab = "重貼現率",
  main = "重貼現率時間序列圖"
)

# 貨幣供給 M2
plot(
  model_data$date,
  model_data$M2,
  type = "l",
  xlab = "年月",
  ylab = "M2",
  main = "貨幣供給時間序列圖"
)

# 消費者物價指數
plot(
  model_data$date,
  model_data$CPI,
  type = "l",
  xlab = "年月",
  ylab = "CPI",
  main = "消費者物價指數時間序列圖"
)

# 建築貸款餘額
plot(
  model_data$date,
  model_data$BL,
  type = "l",
  xlab = "年月",
  ylab = "建築貸款餘額",
  main = "建築貸款餘額時間序列圖"
)

# 恢復單一圖形版面
par(mfrow = c(1, 1))


# ADF 單根檢定 --------------------------------------------------------

# 房價自然對數 ADF 檢定
adf_price <- ur.df(
  model_data$ln_price,
  type = "trend",
  selectlags = "AIC"
)

# 重貼現率 ADF 檢定
adf_rate <- ur.df(
  model_data$RATE,
  type = "trend",
  selectlags = "AIC"
)

# M2 ADF 檢定
adf_m2 <- ur.df(
  model_data$M2,
  type = "trend",
  selectlags = "AIC"
)

# CPI ADF 檢定
adf_cpi <- ur.df(
  model_data$CPI,
  type = "trend",
  selectlags = "AIC"
)

# 建築貸款餘額 ADF 檢定
adf_bl <- ur.df(
  model_data$BL,
  type = "trend",
  selectlags = "AIC"
)

# 輸出 ADF 檢定結果
summary(adf_price)
summary(adf_rate)
summary(adf_m2)
summary(adf_cpi)
summary(adf_bl)


# Johansen 共整合檢定 -------------------------------------------------

# 建立共整合檢定資料
johansen_data <- model_data[, c(
  "ln_price",
  "RATE",
  "M2",
  "CPI",
  "BL"
)]

# Johansen 跡檢定
johansen_test <- ca.jo(
  johansen_data,
  type = "trace",
  ecdet = "const",
  K = 2,
  spec = "transitory"
)

# 輸出共整合檢定結果
summary(johansen_test)


# OLS 模型建構與診斷 --------------------------------------------------

# 建立 OLS 模型
model_lm <- lm(
  ln_price ~ RATE + M2 + CPI + BL,
  data = model_data
)

# OLS 模型摘要
summary(model_lm)

# OLS 模型 AIC
AIC(model_lm)

# OLS 殘差診斷圖
par(mfrow = c(2, 2))
plot(model_lm)
par(mfrow = c(1, 1))


# 未納入時間變數之 GAM 模型 -------------------------------------------

# 建立未納入時間趨勢之 GAM 模型
model_gam <- gam(
  ln_price ~
    s(RATE, k = 4) +
    s(M2, k = 10) +
    s(CPI, k = 10) +
    s(BL, k = 10),
  data = model_data,
  method = "REML"
)

# 模型摘要
summary(model_gam)

# 模型 AIC
AIC(model_gam)

# 模型診斷
gam.check(model_gam)

# 非線性共線性檢查
concurvity(
  model_gam,
  full = TRUE
)

# 平滑函數圖
plot(
  model_gam,
  shade = TRUE,
  pages = 1,
  residuals = FALSE
)


# GAM + time 模型 -----------------------------------------------------

# 建立納入時間趨勢之 GAM 模型
model_gam_time <- gam(
  ln_price ~
    s(RATE, k = 4) +
    s(M2, k = 10) +
    s(CPI, k = 10) +
    s(BL, k = 10) +
    s(time, k = 20),
  data = model_data,
  method = "REML"
)

# 模型摘要
summary(model_gam_time)

# 模型 AIC
AIC(model_gam_time)

# 基底維度與殘差診斷
gam.check(model_gam_time)

# 非線性共線性檢查
concurvity(
  model_gam_time,
  full = TRUE
)

# 繪製平滑函數圖
plot(
  model_gam_time,
  shade = TRUE,
  pages = 1,
  residuals = FALSE,
  seWithMean = TRUE
)


# 三種模型比較 --------------------------------------------------------

# 比較三種模型之 AIC
AIC(
  model_lm,
  model_gam,
  model_gam_time
)

# OLS 調整後決定係數
summary(model_lm)$adj.r.squared

# GAM 調整後決定係數
summary(model_gam)$r.sq

# GAM 偏差解釋比例
summary(model_gam)$dev.expl

# GAM + time 調整後決定係數
summary(model_gam_time)$r.sq

# GAM + time 偏差解釋比例
summary(model_gam_time)$dev.expl


# GAM + time 模型殘差診斷 ---------------------------------------------

# 取得模型殘差及配適值
gam_residuals <- residuals(
  model_gam_time,
  type = "deviance"
)

gam_fitted <- fitted(model_gam_time)

# 殘差與配適值圖
plot(
  gam_fitted,
  gam_residuals,
  xlab = "配適值",
  ylab = "偏差殘差",
  main = "GAM + time 殘差與配適值圖"
)

abline(
  h = 0,
  lty = 2
)

# 常態 Q-Q 圖
qqnorm(
  gam_residuals,
  main = "GAM + time 殘差常態 Q-Q 圖"
)

qqline(gam_residuals)

# 殘差時間序列圖
plot(
  model_data$date,
  gam_residuals,
  type = "l",
  xlab = "年月",
  ylab = "偏差殘差",
  main = "GAM + time 殘差時間序列圖"
)

abline(
  h = 0,
  lty = 2
)

# 殘差自相關圖
acf(
  gam_residuals,
  main = "GAM + time 殘差自相關函數圖"
)
```
